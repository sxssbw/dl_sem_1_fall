# Метрика

Перед решением любой задачи важно понимать, как будет оцениваться ваше решение. В данном случае мы используем стандартную для задачи классификации метрику ROC-AUC. Ее можно вычислить, используя только предсказанные вероятности и истинные классы без конкретного порога классификации + она раотает даже если классы в данных сильно несбалансированны (примеров одного класса в десятки раз больше примеров длугого). Именно поэтому она очень удобна для соревнований.

Посчитать ее легко:


```python
from sklearn.metrics import roc_auc_score

y_true = [
    0,
    1,
    1,
    0,
    1
]

y_predictions = [
    0.1,
    0.9,
    0.4,
    0.6,
    0.61
]

roc_auc_score(y_true, y_predictions)
```

```python
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, RobustScaler, LabelEncoder, OneHotEncoder
from sklearn.pipeline import make_pipeline
```

```python
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

one_hot_encoded = encoder.fit_transform(data[cat_cols])
one_hot_df = pd.DataFrame(
    one_hot_encoded, 
    columns=encoder.get_feature_names_out(cat_cols), 
    index=data.index
)

data_encoded = pd.concat([data, one_hot_df], axis=1)
data_encoded = data_encoded.drop(columns=cat_cols)
data_encoded
```

```python
X = data_encoded.drop(columns=['Churn'])
y = data_encoded['Churn']
```

```python
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.2, shuffle=True, random_state=44)
```

```python
scaler = StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_val_scaled = scaler.transform(X_val)
```

```python
from sklearn.metrics import roc_auc_score

model_cv = LogisticRegressionCV(
    Cs=[100, 10, 1, 0.1, 0.01, 0.001], 
    cv=5, 
    scoring='roc_auc',     
    refit=True,            
    max_iter=1000, 
    class_weight='balanced',
    n_jobs=-1,
    random_state=44
)

model_cv.fit(X_train_scaled, y_train)

print(f"Лучший C: {model_cv.C_[0]}")

y_pred_proba = model_cv.predict_proba(X_val_scaled)[:, 1]
print(f"roc_auc на вале: {roc_auc_score(y_val, y_pred_proba):.4f}")
```

```python
import catboost
```

```python
X_origin = data.drop(columns=['Churn']).copy()
y = data['Churn'].copy()

categorical_indices = [X_origin.columns.get_loc(col) for col in cat_cols]

X_train_origin, X_val_origin, y_train, y_val = train_test_split(X_origin, y,
                                                       train_size=0.8,
                                                       random_state=42)
```

```python
boosting_model = catboost.CatBoostClassifier(n_estimators=200,
                                             cat_features=cat_cols)

boosting_model.fit(X_train_origin, y_train)

y_train_predicted = boosting_model.predict_proba(X_train_origin)[:, 1]
y_val_predicted = boosting_model.predict_proba(X_val_origin)[:, 1]
```

```python
train_auc = roc_auc_score(y_train, y_train_predicted)
val_auc = roc_auc_score(y_val, y_val_predicted)
print(f"Train AUC = {train_auc:.4f}, Val AUC = {val_auc:.4f}")
```

```python
boosting_model_gs = catboost.CatBoostClassifier(
    cat_features=cat_cols,
    eval_metric='AUC',
    logging_level='Silent',
    random_state=42
)

param_grid = {
    'iterations': [200, 500, 800],
    'learning_rate': [0.01, 0.05, 0.1]
}

boosting_model_gs.grid_search(
    param_grid,
    X_train_origin,
    y_train,
    plot=False, 
    refit=True,
    verbose=False
)

y_train_pred_gs = boosting_model_gs.predict_proba(X_train_origin)[:, 1]
y_val_pred_gs = boosting_model_gs.predict_proba(X_val_origin)[:, 1]

train_auc_gs = roc_auc_score(y_train, y_train_pred_gs)
val_auc_gs = roc_auc_score(y_val, y_val_pred_gs)

best_params = boosting_model_gs.get_all_params()
print(f"Лучшие параметры: n_estimators = {best_params['n_estimators']}, learning_rate = {best_params['learning_rate']}")
print(f"После подбора: Train AUC = {train_auc_gs:.4f}, Valid AUC = {val_auc_gs:.4f}")
```
